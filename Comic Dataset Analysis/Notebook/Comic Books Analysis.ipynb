```python
# 1. Which genres generate the highest audience engagement?

genre_perf=df.groupby('Genre')['Rating (out of 10)'].mean().reset_index()
genre_perf=genre_perf.sort_values('Rating (out of 10)',ascending=False)

fig=px.bar(
    genre_perf,
    x='Genre',
    y='Rating (out of 10)',
    color='Rating (out of 10)',
    color_continuous_scale='Viridis'
)

fig.update_layout(title='Top Performing Genres by Audience Rating')
fig.show()
```

# Top Performing Genres by Audience Rating

1. **Adventure/Action Dominates the Chart** : Adventure/Action leads all genres with a near-perfect rating close to 9.8/10,suggesting audiences strongly favour high-energy, escapist content over slower-paced or niche storytelling formats.

2. **Action-Hybrid Genres Sweep the Top 4** : The top 4 spots are all Action-blended genres (Adventure/Action,Historical/Action, Action/Mystery, Action/Sci-Fi), confirming that pairing Action with any secondary genre is a reliable formula for high audience satisfaction.

3. **Sharp Drop-Off After the Top 4** : Ratings fall noticeably from ~9.3 (top 4) to ~8.3 at rank 5 (Action/Supernatural), revealing a two-tier structure — a high-performing elite cluster and a large mid-tier pack bunched tightly between 8.0–8.3.

4. **Mid-Tier Genres Show Minimal Differentiation** : From rank 5 onward, roughly 20+ genres share ratings between 8.0 and 8.3  a spread of just 0.3 points. This plateau suggests audience perception of these genres is nearly interchangeable, making brand, cast, and production quality the real differentiators at this level.

5. **Niche and Drama-Heavy Genres Lag at the Tail** : Genres like Autobiography/Drama, Fantasy/Drama, and Literary/Drama anchor the bottom of the chart at ~7.8–8.0. Their lower ratings likely reflect a narrower audience base and higher expectation thresholds — viewers who seekout these genres tend to be more critical consumers.

```python
# 2. How has comic production evolved over the years?

yearly=df.groupby('Release Year').size().reset_index(name='Total Comics')
yearly=yearly.sort_values('Release Year')

fig=px.line(
    yearly,
    x='Release Year',
    y='Total Comics',
    markers=True,
    color_discrete_sequence=['#e63946']
)

fig.update_layout(title='Comic Production Trend Over Time')
fig.show()
```

# Comic Production Trend Over Time Insights

1. **Stagnant Era from 2000 to 2008** : Production remained flat between ~210–245 comics annually for nearly a decade, suggesting the industry was in a consolidation phase with limited appetite for expanding output during the early 2000s.

2. **Explosive Growth Phase from 2009 to 2023** : Output nearly tripled — from ~245 to a peak of ~595 — over just 14 years.This sustained surge likely reflects the global boom in comic adaptations,streaming demand for source material, and the rise of webtoon and digital publishing platforms.

3. **All-Time Peak Reached in 2023** : Production hit its highest recorded point at approximately 595 comics in 2023,marking the climax of a decade-long growth run and setting a benchmark that   no prior year came close to matching.

4. **Sharp Decline from 2024 to 2026** : Output dropped steeply from ~595 (2023) to ~450 (2026) — a fall of roughly
   145 units in just three years. This rapid contraction may signal market saturation, reduced platform investment, or a post-pandemic content correction after years of overproduction.

5. **The 2026 Dip Warrants Close Monitoring** : At ~450 comics, 2026 production is back to 2015 levels, erasing over a decade of growth momentum. Whether this is a temporary correction or the start of a structural decline is the critical question — the next 2–3 data points will be decisive in determining the industry's trajectory.

```python
# 3. Which age groups receive the highest ratings?

age=df.groupby('Age Rating')['Rating (out of 10)'].mean().reset_index()

fig=px.funnel(
    age,
    x='Rating (out of 10)',
    y='Age Rating',
    color='Age Rating',
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_layout(title='Ratings Across Age Categories')
fig.show()
```

# Ratings Across Age Categories

1. **Young Adult Leads but by a Razor-Thin Margin** : Young Adult tops the chart at 8.078/10, but the gap between first and last place (Mature at 8.040) is just 0.038 points — indicating that age rating has virtually no meaningful impact on how audiences score comic content.

2. **All Ages Ranks Second — Defying Expectations** : With a rating of 8.075, All Ages content nearly matches Young Adult,suggesting that universally accessible comics are crafted with consistently high quality and appeal broadly enough to earn strong scores across reviewer demographics.

3. **Mature Content Does Not Translate to Higher Ratings** : Both Mature (8.040) and Mature 17+ (8.051) rank at the bottom two positions,challenging the assumption that adult-oriented, complex content automatically earns more critical appreciation from audiences.

4. **The Entire Rating Band Spans Only 0.038 Points** : All five age categories sit within an extraordinarily tight range of 8.04 to 8.08. This near-identical clustering suggests that story quality,art, and execution — not age classification — are the true drivers of audience satisfaction.

5. **Teen+ Holds a Competitive Middle Ground** : Teen+ earns 8.062, placing it solidly in the middle of the pack. This category likely benefits from a large, highly engaged readership base and genre diversity (action, fantasy, romance), which together stabilise its average rating close to the overall group mean.

```python
# 4. Which countries produce the highest-rated comics?

country=df.groupby('Country of Origin')['Rating (out of 10)'].mean().reset_index()
country=country.sort_values('Rating (out of 10)',ascending=False)

fig=px.bar(
    country,
    x='Country of Origin',
    y='Rating (out of 10)',
    color='Country of Origin',
    color_discrete_sequence=px.colors.qualitative.Bold
)

fig.update_layout(title='Country-wise Average Ratings')
fig.show()
```

# Country-wise Average Ratings

1. **Franco-Iranian and Franco-Spanish Collaborations Lead the Chart** : France/Iran and France/Spain claim the top two spots with ratings of ~9.1 and ~9.0 respectively, suggesting that Franco-international co-productions carry a distinctive artistic quality — likely driven by auteur-style storytelling traditions that resonate strongly with global audiences.

2. **East Asian Origins Dominate the Upper-Mid Tier** : Japan, South Korea/Japan, South Korea, and South Korea/USA all cluster between 8.1–8.2, confirming that East Asian comic industries consistently deliver high-quality content. The presence of collaborative entries (South Korea/Japan, South Korea/USA) suggests cross-border co-productions amplify rating performance further.

3. **Western Nations Form a Tightly Packed Mid-Tier** : USA, New Zealand, Australia, UK, France, China, Canada, Germany, and Sweden all sit within a narrow 7.9–8.0 band. Despite vast differences in industry size and cultural tradition, these countries converge on nearly identical audience ratings — pointing to a global quality baseline for comics.

4. **Tail-End Countries Are Still Competitive, Not Poor Performers** : Belgium, Switzerland, Spain, Netherlands, and France/Belgium anchor the bottom at ~7.7–7.8 — but these are not weak scores. A 7.7 out of 10 remains a strong rating, indicating even the lowest-ranked origins produce content that audiences genuinely appreciate.

5. **Co-Production Entries Consistently Outperform Solo-Nation Entries** : Across the chart, collaborative origin tags (France/Iran, France/Spain, South Korea/Japan, South Korea/USA) tend to rank above their individual country counterparts. This pattern implies that cross-cultural creative partnerships produce a measurable lift in perceived quality and audience satisfaction.

```python
# 5. Which comic formats are most popular?

formats=df['Format'].value_counts().reset_index()
formats.columns=['Format','Count']

fig=px.pie(
    formats,
    names='Format',
    values='Count',
    hole=0.4,
    color_discrete_sequence=px.colors.sequential.RdBu
)

fig.update_layout(title='Comic Format Popularity')
fig.show()
```

# Comic Format Popularity

1. **Japanese Formats Collectively Dominate the Market** : Tankobon (9.06%), Digital Manga (9.06%), and Manga Volume (8.93%) together account for over 27% of all comic formats — making Japanese-origin formats the single largest format bloc by a wide margin and reflecting manga's overwhelming global reach and reader loyalty.

2. **No Single Format Commands a Clear Majority** : The largest individual share is just 9.06%, shared by both Tankobon and Digital Manga. This near-equal fragmentation across 16+ formats indicates a highly diversified market where reader preferences are spread broadly rather than concentrated around one dominant delivery format.

3. **Digital Formats Are a Significant and Growing Force** : Digital Manga (9.06%), Digital Single Issue (5.88%), Digital Webtoon (4.35%), and Digital Manhua together represent a substantial combined share, signalling that digital distribution is not a niche channel — it is a mainstream and structurally embedded part of how comic content is consumed globally.

4. **Webtoon Ecosystem Commands a Meaningful Combined Share** : Webtoon (4.11%), Webtoon Chapter Collection (4.18%), and Digital Webtoon (4.35%) collectively account for ~12.6% of the market. This consolidated webtoon presence reflects the rapid global rise of the vertical-scroll format,driven largely by mobile-first audiences in Asia and beyond.

5. **Traditional Print Formats Remain Resilient** : Graphic Novel (7.64%), Trade Paperback (6.95%), Single Issue (6.24%), and Hardcover (5.99%) still hold a combined share of over 26%, demonstrating that physical print formats retain strong reader demand despite the surge in digital and platform-based alternatives.

The democratisation of digital color tools likely enables more creators to publish, but with uneven results — raising the ceiling and lowering the floor simultaneously.

5. **Watercolor and Pastel Styles Produce Tight, Reliable Mid-Range Ratings** : Both Watercolor and Full Color (Pastel) show compact, symmetrical violin shapes with IQRs tightly packed around 7.8–8.2. These niche artistic styles attract a selective audience with calibrated expectations, resulting in consistently solid but rarely polarising scores.

```python
# 7.Which genre-country combinations dominate the global comic market?

import matplotlib.pyplot as plt
import squarify
import seaborn as sns
import pandas as pd

country_genre_counts = df.groupby(['Country of Origin', 'Genre']).size().reset_index(name='Count')

top_n = 30
top_counts = country_genre_counts.sort_values('Count', ascending=False).head(top_n)

top_counts['label'] = top_counts['Genre'] + "\n" + top_counts['Count'].astype(str)

unique_genres = top_counts['Genre'].unique()
palette = sns.color_palette("Set3", len(unique_genres))
genre_color_map = dict(zip(unique_genres, palette))
colors = top_counts['Genre'].map(genre_color_map)

plt.figure(figsize=(12, 7))
squarify.plot(
    sizes=top_counts['Count'],
    label=top_counts['label'],
    color=colors,
    alpha=0.8,
    text_kwargs={'fontsize':9}
)
plt.axis('off')
plt.title("Top 30 Genre Distribution Across Countries", fontsize=14)

import matplotlib.patches as mpatches

patches = [mpatches.Patch(color=color, label=genre) for genre, color in genre_color_map.items()]
plt.legend(handles=patches, bbox_to_anchor=(1, 1), loc='upper left', borderaxespad=0.)
plt.tight_layout()
plt.show()
```

# Top 30 Genre Distribution Across Countries

1. **Superhero/Thriller leads at 94** : the single most globally distributed genre, confirming superhero content with dark, suspenseful undertones has the broadest cross-country appeal.

2. **Superhero/Action follows closely at 91** : action remains the most universally marketable pairing for superhero narratives across diverse cultural markets.

3. **Superhero genres dominate 8 of the top 30 slots** : reflecting the genre's unmatched global franchise infrastructure and consistent international audience demand.

4. **Superhero/Psychological at 82 outranks pure action blends** : audiences worldwide show strong appetite for mentally complex, character-driven superhero storytelling.

5. **Non-superhero genres cap out at 68 (Mecha/Sci-Fi)** : highlighting a clear ceiling for niche genres versus the mainstream dominance of the superhero category globally.

```python
# 8.Which metrics are highly correlated?

import plotly.express as px
corr=df[['Page Count','Release Year',
         'Rating (out of 10)','Volume Count']].corr()

fig=px.imshow(
    corr,
    text_auto=True,
    color_continuous_scale='RdBu'
)

fig.update_layout(title='Correlation Between Business Metrics')
fig.show()
```

# Correlation Between Business Metrics

1. **Page Count & Volume Count share the strongest relationship (0.42)** : longer comics tend to span more volumes, confirming that content depth and series length scale together as a natural publishing pattern.

2. **Release Year is virtually uncorrelated with everything** : its highest correlation is just 0.13 with Page Count, meaning newer comics are neither longer, better rated, nor more voluminous than older ones.

3. **Rating is statistically independent of all other metrics** : correlations of 0.16, 0.03, and 0.15 confirm that page count, recency, and volume count have no meaningful influence on how audiences score a comic.

4. **Release Year & Rating produce the weakest link in the matrix (0.03)** : definitively ruling out recency bias, meaning audiences do not rate newer comics higher simply because they are new.

5. **No strong correlations exist anywhere in this matrix** : with a maximum of just 0.42, all business metrics operate largely independently, signalling that comic success is driven by qualitative factors outside these four measurable dimensions.

```python
# Outliers Detection
from scipy.stats import zscore

df['zscore_rating'] = zscore(df['Rating (out of 10)'])

outliers = df[df['zscore_rating'].abs() > 3]
print(outliers[['Title', 'Rating (out of 10)']])
```

```python
# IQR Method 

Q1 = df['Rating (out of 10)'].quantile(0.25)
Q3 = df['Rating (out of 10)'].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df['Rating (out of 10)'] < Q1 - 1.5 * IQR) |
    (df['Rating (out of 10)'] > Q3 + 1.5 * IQR)
]

print(outliers[['Title', 'Rating (out of 10)']])
```
