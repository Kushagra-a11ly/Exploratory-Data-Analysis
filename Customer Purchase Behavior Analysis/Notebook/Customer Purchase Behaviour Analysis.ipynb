```python
# Does age influence how much customers spend?

sns.scatterplot(data=df,
                x='Age',
                y='Purchase Amount (USD)',
                color="#2A9D8F",
                alpha=0.6)

sns.regplot(data=df,
            x='Age',
            y='Purchase Amount (USD)',
            scatter=False,
            color="#E76F51")

plt.title("Age vs Spending Trend")
plt.show()
```

```python
# Insights:

# 1. Age has no predictive power over spending. The regression line is virtually flat at ~$60, meaning age tells you nothing about how much a customer will spend.
# 2. Spending range is identical across all age groups. Every cohort from 18 to 70 spends within the same $20–$100 band, with no group consistently spending more or less than another.
# 3. The 25–45 age band has the highest customer concentration. Point density is visibly thicker in this range, making it the most important segment for volume-driven decisions like inventory and acquisition budgets.
# 4. The $20–$100 spending boundaries are suspiciously clean. Real-world data rarely cuts off this sharply, suggesting a hard pricing tier in the catalogue or data truncation that needs investigation before deeper analysis is done.
# 5. The real drivers of spending are hidden elsewhere. With age explaining virtually no variance, factors like income, product category, or geography are almost certainly the true predictors — a multivariate model is the logical next step.
# 6. Older customers (60–70) are under-represented but spend equally. Despite appearing far less frequently in the data, this age group spends at exactly the same level as younger customers, signalling an underserved segment with genuine growth potential.
```

```python
# Do discounts increase or decrease how much customers spend per transaction?

sns.barplot(data=df,
            x='Discount Applied',
            y='Purchase Amount (USD)',
            palette="coolwarm")

plt.title("Effect of Discount on Purchase Amount")
plt.show()
```

```python
# Insights:

# 1. Discounts have virtually no effect on purchase amount. Both groups spend almost identically — ~$59 with a discount and ~$60 without — making the discount programme commercially questionable.
# 2. The discount is actually associated with slightly lower spending. Customers who received a discount spent marginally less than those who didn't, which is the opposite of what a well-designed discount strategy should achieve.
# 3. Error bars are extremely tight, meaning this result is statistically reliable. The narrow confidence intervals on both bars indicate low variance, so this near-zero difference is not a fluke — it is a stable pattern in the data.
# 4. The discount programme is likely not driving basket size. If discounts were encouraging customers to buy more or trade up, the "Yes" bar would be meaningfully taller — the flat result suggests customers are simply buying what they intended to regardless.
# 5. The business may be giving away margin for no incremental gain. Since discounted customers spend the same or less, every discount applied is pure margin erosion without a corresponding lift in revenue or basket value.
# 6. The effectiveness of discounts should be measured on purchase frequency, not amount. This chart only shows spend per transaction — discounts might still be driving repeat visits or customer acquisition, which would require a separate cohort or time-series analysis to evaluate properly.
```

```python
# Do promo codes actually change how much customers spend?

sns.boxplot(data=df,
            x='Promo Code Used',
            y='Purchase Amount (USD)',
            palette="pastel")

plt.title("Promo Code Usage Impact")
plt.show()
```

```python
# Insights:

# 1. Promo codes have no meaningful impact on purchase amount. Both groups share an almost identical median of ~$60, indicating that using a promo code does not push customers to spend more per transaction.
# 2. The spread of spending is virtually the same with or without a promo code. Both boxes span roughly $40–$80 with identical whiskers from $20 to $100, meaning the entire distribution of spend is unaffected by promo code usage.
# 3. The "No promo" group has a very slightly higher Q3, suggesting marginal upward skew. The upper quartile of non-promo customers reaches just above $80 compared to exactly $80 for promo users, hinting that organic buyers occasionally spend a touch more at the higher end.
# 4. Promo codes are not converting low spenders into higher spenders. If promos were effective at upselling, the lower quartile of the "Yes" group would be elevated — instead it sits at the same ~$40 as non-users, showing no lift at the bottom of the distribution either.
# 5. The business is absorbing promo code costs with zero basket size return. Since spend distribution is identical across both groups, every promo code redeemed represents a direct margin cost with no compensating increase in purchase value.
# 6. Promo code strategy needs to be re-evaluated using frequency and retention metrics. Like discounts, promos may serve a purpose in driving repeat purchases or new customer acquisition — but this chart confirms they are doing nothing to increase how much customers spend in a single visit.
```

```python
# What is the share of each payment method in total transactions?

df['Payment Method'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=sns.color_palette("Set3"),
    startangle=90
)

plt.title("Payment Method Distribution")
plt.ylabel("")
plt.show()
```

```python
# Insights:

# 1. Payment method preference is remarkably uniform across all options. All six methods fall within a razor-thin 1.6 percentage point range (16.2%–17.8%), which is statistically near-perfect equality across options.
# 2. Credit card leads marginally but holds no dominant position. At 17.8% it is the most used method, but the gap over the lowest (Bank Transfer and Debit Card at 16.2%) is too small to call it a customer favourite in any meaningful sense.
# 3. This distribution strongly suggests the data may be synthetic or heavily sampled. Real-world payment preferences almost never distribute this evenly — typically card payments dominate heavily, making this uniformity a red flag worth flagging to the data team.
# 4. The presence of both Cash and Venmo at comparable levels is commercially unusual. Cash and digital wallets like Venmo typically serve very different customer demographics, yet both sit at ~16.6–16.7%, suggesting either an unusually diverse customer base or a data aggregation issue.
# 5. No single payment method requires prioritisation in checkout optimisation. Since no method commands a dominant share, engineering and UX resources should not be skewed toward optimising any one payment flow over another.
# 6. The even split makes it impossible to draw behavioural conclusions about customer segments. Payment method choice often correlates with age, income, or digital literacy — but with no method standing out, segmentation analysis using payment type as a variable would yield no actionable signal.
```

```python
# How does revenue vary by category across different seasons?

pivot = df.pivot_table(values='Purchase Amount (USD)',
                       index='Category',
                       columns='Season',
                       aggfunc='sum')

sns.heatmap(pivot,
            annot=True,
            fmt=".0f",
            cmap="YlGnBu",
            linewidths=0.5)

plt.title("Seasonal Revenue Heatmap")
plt.show()
```

```python
# Insights:

# 1. Clothing is the dominant revenue category by a significant margin. Across all four seasons it generates $23K–$28K, sitting in a completely different tier from every other category and representing the clear commercial backbone of the business.
# 2. Outerwear is chronically underperforming, even in its natural season. Despite winter being its logical peak, Outerwear generates only $4,562 in winter — barely more than its summer low of $4,278 — suggesting weak product range, poor visibility, or chronic underpricing.
# 3. Clothing's summer dip is the only notable seasonal pattern in the data. Revenue drops from $27,692 in spring to $23,078 in summer before recovering to $27,274 in winter, indicating a mid-year demand softness that could be addressed with targeted summer promotions.
# 4. Accessories maintain surprisingly stable revenue across all seasons. Ranging narrowly from $17,007 to $19,874, Accessories show almost no seasonal sensitivity — making them a reliable, low-risk revenue contributor ideal for consistent marketing investment year-round.
# 5. Footwear occupies a stagnant middle tier with no seasonal breakout. Revenue hovers between $8,480 and $9,555 regardless of season, indicating either a category with a hard ceiling in the current catalogue or one that lacks seasonal product refreshes to drive demand spikes.
# 6. Spring is the strongest overall season when summing across all categories. Clothing peaks in spring, Footwear is at its highest, and Accessories posts its second-lowest — making spring the most important trading period and the logical moment to concentrate promotional and inventory investment.
```

```python
# Do repeat customers spend more?

sns.scatterplot(data=df,
                x='Previous Purchases',
                y='Purchase Amount (USD)',
                color="#8E44AD",
                alpha=0.6)

plt.title("Repeat Purchases vs Spending")
plt.show()
```

```python
# Insights:

# 1. Purchase history has zero influence on how much a customer spends. The scatter is completely uniform from 0 to 50 previous purchases, with no upward or downward trend — loyalty and spend per transaction are entirely disconnected variables.
# 2. This directly challenges the common assumption that loyal customers spend more. A customer with 50 previous purchases spends no more on average than a first-time buyer, which undermines any business case for loyalty programmes designed to increase basket size.
# 3. Spend is randomly distributed between $20 and $100 regardless of purchase frequency. Just like the age scatter seen earlier, the $20–$100 hard boundary reappears here, further confirming this is a dataset-level constraint rather than a natural behavioural pattern.
# 4. There is a healthy volume of highly loyal customers in this dataset. A substantial number of customers have 40–50 previous purchases, indicating strong retention — but since they spend the same as new customers, the value of retention lies in frequency, not transaction size.
# 5. New customers (0–5 purchases) show the same spending range as veterans. There is no onboarding effect or honeymoon spend spike from new customers, suggesting first purchases are not experimentation-driven and customers arrive with a relatively fixed budget in mind.
# 6. Upselling and cross-selling strategies cannot rely on purchase history as a trigger. Since repeat purchase count predicts nothing about spend level, personalisation engines and upsell prompts need to be driven by category affinity, recency, or basket composition — not loyalty tier alone.
```

```python
# Insights:

# 1. Subscription status has no effect whatsoever on purchase amount. Both groups share an identical median of ~$60 and near-identical box shapes, meaning subscribers and non-subscribers spend exactly the same per transaction.
# 2. This is now the third variable in a row — after discounts and promo codes — that fails to move spending. The consistent pattern across all three charts strongly suggests that no single promotional or loyalty mechanism in this business is successfully driving basket size uplift.
# 3. The upper quartile of non-subscribers is fractionally higher, mirroring the promo code chart pattern. Non-subscribed customers edge slightly above $80 at Q3 versus exactly $80 for subscribers — a negligible but recurring theme where the "incentivised" group paradoxically spends marginally less.
# 4. Subscribers are not being converted into higher-value customers through the programme. A well-functioning subscription model should gradually shift members toward larger baskets through exclusive offers or member pricing — the flat distribution here shows this is not happening.
# 5. The subscription programme's value must be justified entirely through purchase frequency, not spend per visit. If subscribers buy more often even at the same transaction value, the programme may still be commercially viable — but that case needs to be made with a frequency or CLV analysis, not spend data.
# 6. Taken together with all prior charts, the data points to a highly commoditised, price-anchored customer base. Age, loyalty, discounts, promos, and subscriptions all fail to shift spending — suggesting customers have a strong, fixed mental price point around $60 that no current intervention is breaking through.
```

```python
# What does our customer age distribution look like?

import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("white")  # removes grid
sns.histplot(df['Age'], bins=10, kde=True, color='teal')
plt.title('Age Distribution of Customers')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```
