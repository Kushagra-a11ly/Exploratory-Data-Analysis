```python
# What is the overall pricing landscape of EVs?

plt.figure(figsize=(8,5))
sns.histplot(df['price_usd'], kde=True, bins=30)
plt.title("Distribution of EV Prices")
plt.xlabel("Price (USD)")
plt.ylabel("Frequency")
plt.show()
```

# Distribution of EV Prices — Key Insights

**1. Right-Skewed Distribution** : The price distribution is positively skewed, with the majority of EVs concentrated in the $40K–$90K range and a long tail extending beyond $200K.

**2. Market Sweet Spot at ~$65K–$75K** : The distribution peaks around $65,000–$75,000, indicating this is the most competitive and densely populated price band in the EV market.

**3. Mass-Market Dominance Below $100K** : Over 75% of EV listings fall under $100,000, confirming the market is predominantly driven by accessible, consumer-grade electric vehicles rather than luxury models.

**4. Thin Premium Segment Beyond $150K** : Frequency drops sharply after $150,000, with very few vehicles priced above $200K — signalling a niche but present ultra-premium EV segment.

**5. Entry-Level Scarcity Under $30K** : Low frequency below $30,000 suggests that truly budget-friendly EVs remain underrepresented, pointing to an affordability gap in the lower price tier.

```python
# Which brands dominate the dataset?

sns.countplot(
    y='brand',
    data=df,
    order=df['brand'].value_counts().index,
    hue='brand',
    palette='viridis',
    legend=False
)
plt.title("Number of Vehicles by Brand")
plt.show()
```

# Number of EV Vehicles by Brand — Key Insights


**1. Tesla's Commanding Lead** : Tesla dominates the dataset with ~430 vehicles — nearly 55% more than second-placed BYD (~280), underlining its unrivalled market presence in the EV space.

**2. BYD & Volkswagen Anchor the Second Tier** : BYD (~280) and Volkswagen (~200) form a clear runner-up cluster, reflecting the rapid global scaling of Chinese and European EV manufacturers.

**3. Tight Mid-Pack Competition** : Kia, Hyundai, BMW, GM/Chevrolet, and Ford all sit within a narrow 130–165 range, indicating intense, evenly matched competition in the mainstream EV segment.

**4. Sharp Drop-Off After the Top 10** : Inventory falls steeply from Mercedes (~100) onward, with Toyota, Rivian, Fisker, and NIO all below 40 — highlighting a long tail of low-volume or emerging players.

**5. Niche Brands Hold Minimal Share** : Porsche, Volvo, Polestar, Xiaomi, Lucid, and Honda each register under 20 vehicles, reflecting early-stage EV commitments or highly selective market positioning.

```python
# Does battery size determine range?

plt.figure(figsize=(7,5))
sns.scatterplot(
    x='battery_capacity_kwh',
    y='range_miles',
    data=df,
    hue='market_segment',
    palette='Set2'
)
plt.title("Battery Capacity vs Range")
plt.show()
```

# Battery Capacity vs Range — Key Insights

**1. Strong Positive Correlation** : Battery capacity and range share a near-linear relationship across all segments — larger kWh consistently delivers higher mileage, confirming capacity as the primary range driver.

**2. All Four Segments Span the Full Capacity Spectrum** : Luxury, Mid-range, Premium, and Budget vehicles are interspersed across 40–120 kWh with no segment exclusively owning a capacity tier, indicating segment pricing is driven by factors beyond battery size alone.

**3. Visible Vertical Spread at Every Capacity Level**  At any given kWh value, range varies by up to 50–60 miles across vehicles, pointing to aerodynamics, drivetrain efficiency, and weight as significant secondary range determinants.

**4. Premium Segment Shows the Widest Range Variance** : Blue (Premium) dots display the most scatter above and below the trend, suggesting premium EVs prioritise performance and features over range consistency.

**5. Budget EVs Cluster at Lower Capacity but Remain Competitive on Range** : Pink (Budget) dots appear mostly between 40–80 kWh yet achieve ranges comparable to higher-capacity peers, hinting at strong efficiency optimisation in budget-tier engineering.

```python
# Do faster cars cost more?

plt.figure(figsize=(7,5))
sns.scatterplot(
    x='acceleration_0_60_mph',
    y='price_usd',
    data=df,
    hue='body_type',
    palette='coolwarm'
)
plt.title("Acceleration vs Price")
plt.show()
```

# Acceleration vs Price — Key Insights

**1. Faster doesn't always mean pricier**:You'd expect a 3-second 0–60 car to cost a fortune, but the scatter tells a different story — high-priced vehicles appear at almost every acceleration level, meaning speed alone isn't what's driving the price tag.

**2. The real money is parked between 3–5 seconds**:That's where the $150K–$260K outliers live — mostly Trucks and Sedans — suggesting buyers in that bracket are paying for performance *and* prestige, not just raw acceleration numbers.

**3. Most of the market clusters comfortably under $100K**:Regardless of body type, the bulk of vehicles across all acceleration bands sit in the $25K–$100K zone, which is exactly where everyday buyers are shopping.

**4. Slower EVs haven't given up on price**:Even in the 7–8 second 0–60 range, you still see vehicles pushing $150K+ — those are likely large SUVs and Vans where size, range, and luxury features matter far more than acceleration.

**5. Body type blends together at every speed**:Trucks, SUVs, Sedans, Coupes, and Hatchbacks are all mixed across the full acceleration spectrum with no clean separation — confirming that body style and speed are independent choices buyers make based on lifestyle, not performance preference alone.

```python
# What features influence each other?

plt.figure(figsize=(10,8))
corr = df.corr(numeric_only=True)
sns.heatmap(
    corr,
    annot=True,
    cmap='Spectral',
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Matrix")
plt.show()
```

# Correlation Matrix — Key Insights

**1. Battery and range are basically the same thing** : With a correlation of 0.98, battery capacity and range miles move together so tightly that using both in a model would be redundant — pick one, because they're telling you the exact same story.

**2. More horsepower means a faster car, not a more expensive one** : Horsepower and torque are nearly perfectly correlated (0.99), and both strongly predict top speed — but their link to price is only moderate (0.76), meaning buyers aren't purely paying for raw power.

**3. Acceleration and top speed are opposites of the same coin** : The -0.97 between acceleration (0–60 time) and top speed makes perfect sense — a lower 0–60 number means the car is faster, so of course it also has a higher top speed. That deep red cell is just physics.

**4. Price has more to do with performance than any single spec** : The strongest price correlations are horsepower (0.76), torque (0.76), and range (0.76) — buyers are paying for the overall performance package, not just one feature in isolation.

**5. Most features are surprisingly independent of each other** : Outside the obvious performance cluster, the vast majority of cells hover near zero — safety rating, seating capacity, weight, and warranty years barely influence price or each other, which is good news for building a clean predictive model.

```python
# Which brands are premium vs budget?

plt.figure(figsize=(10,6))
sns.boxplot(
    x='price_usd',
    y='brand',
    data=df,
    hue='brand',
    palette='pastel',
    legend=False
)
plt.title("Price Distribution by Brand")
plt.show()
```

# Price Distribution by Brand — Key Insights

**1. Lucid and Porsche are in a league of their own** : Both brands sit with medians well above $150K and whiskers stretching past $250K — they're not competing with the rest of the market, they're competing with each other in the ultra-luxury tier.

**2. Mercedes plays premium but keeps one foot in the mainstream** : Its box spans roughly $100K–$175K with a wide spread, meaning Mercedes buyers get very different cars at very different prices — the brand stretches further than any other outside the ultra-luxury two.

**3. Toyota, Hyundai, Kia, and BYD are fighting the same battle** : Tight boxes clustered between $30K–$80K with minimal whisker spread — these brands know exactly who they're selling to and price consistently, leaving little room for surprise.

**4. Tesla's box is deceptively wide** : Spanning roughly $70K–$130K+, Tesla covers more ground than most brands in a single lineup — the Model 3 and the Model S are worlds apart, and that spread tells the whole story.

**5. Honda is barely in the conversation** : The narrowest, leftmost box on the chart tells you Honda's EV presence is minimal, tightly priced, and still finding its footing — a stark contrast to its dominance in the traditional auto market.

```python
# Which vehicle types offer better range?

plt.figure(figsize=(8,5))
sns.boxplot(
    x='body_type',
    y='range_miles',
    data=df,
    hue='body_type',
    palette='muted',
    legend=False
)
plt.title("Range by Body Type")
plt.show()
```

# Range by Body Type — Key Insights

**1. Body type barely moves the needle on range** : Every single box sits with a median right around 255–270 miles — if you're choosing a body type hoping to squeeze out more range, the data says you're optimising the wrong variable.

**2. Coupe edges out the rest, but only just** : With the highest median and a box that sits slightly higher than its peers, Coupes have a marginal range advantage — likely because they're lighter and more aerodynamic, but it's not a gap worth getting excited about.

**3. The spread is what's interesting, not the middle** : All six body types reach up to 440–450 miles at the top and down to ~150 miles at the bottom — the ceiling and floor are identical, meaning the best and worst range vehicles exist in every body style.

**4. Vans are more competitive than their reputation suggests** : Conventionally seen as gas-guzzlers, Vans here match SUVs and Sedans almost perfectly on median range — a sign that EV engineering is quietly closing the gap even in traditionally inefficient form factors.

**5. Trucks show the widest IQR, meaning the most unpredictable range** : The Truck box is the tallest of the group — buy a truck and your range experience could be anywhere from underwhelming to genuinely impressive, depending heavily on which model and spec you land on.

```python
# Is EV adoption increasing?

sales_trend = df.groupby('year')['annual_sales_units'].sum()

plt.figure(figsize=(8,5))
plt.plot(
    sales_trend.index,
    sales_trend.values,
    marker='o',
    color='teal'
)
plt.title("EV Sales Trend Over Time")
plt.xlabel("Year")
plt.ylabel("Units Sold")
plt.show()
```

# EV Sales Trend Over Time — Key Insights

**1. Six years of growth wiped out in one** : The market climbed steadily from under 5M units in 2020 all the way to ~88M in 2025 — then a single year erased more than half that volume, which is the kind of reversal that demands explanation before drawing any conclusions.

**2. The 2020–2025 growth curve is genuinely remarkable** : Units sold roughly doubled every 18 months across that stretch — that's not organic market growth, that's an industry in full expansion mode driven by new model launches, government incentives, and falling battery costs all hitting at once.

**3. 2025 was the peak, but it may be a ceiling not a cliff** : At ~88M units, 2025 represents the all-time high — whether the 2026 drop reflects market saturation, policy changes, subsidy rollbacks, or simply a dataset cutoff mid-year is the critical question this chart alone can't answer.

**4. The 2026 figure should be treated with caution** : A drop to ~28M in a single year would be historically unprecedented for a growing industry — it's far more likely this reflects partial-year data or a reporting gap than a genuine market collapse.

**5. The underlying trajectory still points upward** : Even if 2026 is taken at face value, 28M units is still nearly triple 2022's volumes — the EV market, even at its post-peak low, is operating at a scale that would have seemed impossible just four years ago.

```python
# How do multiple features interact together?

sns.pairplot(
    df[['price_usd','range_miles','battery_capacity_kwh','charging_speed_kw']],
    corner=True,
    plot_kws={'color': 'purple', 'alpha': 0.6}
)
plt.show()
```

# Pair Plot — Price, Range, Battery & Charging Speed — Key Insights

**1. Battery and range are inseparable — and this plot proves it visually** : That near-perfect diagonal line in the battery vs. range scatter is one of the cleanest relationships you'll see in any EV dataset — if you know the battery size, you essentially already know the range.

**2. Price is right-skewed and the rest of the market pays for it** : The price histogram tells the same story as before — most vehicles pile up under $100K, with a long tail of expensive outliers dragging the average up and making price the hardest variable to predict cleanly.

**3. Charging speed has almost nothing to do with anything else** : Every scatter involving charging_speed_kw looks like someone threw confetti — no meaningful pattern with price, range, or battery capacity, which is surprising and suggests charging infrastructure decisions are made independently of vehicle specs.

**4. Range distribution has an unusual bimodal hint** : The range histogram peaks early around 150–175 miles, dips, then rises again toward 275–300 miles — there may be two distinct buyer tiers being served here: urban short-range commuters and long-distance drivers.

**5. Battery capacity clusters between 60–100 kWh with no dominant size** : The battery histogram is relatively flat across that band rather than spiking at one size — manufacturers aren't converging on a single standard pack size yet, which means the market is still figuring out what "enough" battery actually means.

```python
# Which segment offers best value?

df['price_per_mile'] = df['price_usd'] / df['range_miles']

plt.figure(figsize=(8,5))
sns.boxplot(
    x='market_segment',
    y='price_per_mile',
    data=df,
    color='lightcoral'
)
plt.title("Value for Money by Segment")
plt.show()
```

# Value for Money by Segment — Key Insights

**1. Budget EVs genuinely deliver the best value per mile — and it's not close** : With a median price-per-mile around $150 and the tightest box of all four segments, Budget vehicles offer the most range for your dollar, consistently and predictably, with almost no variance in that promise.

**2. Luxury buyers are paying a serious premium for every mile they drive** : The Luxury median sits near $480/mile — over three times the Budget segment — and that gap is the literal cost of the brand, the interior, and the badge, because the range itself isn't proportionally better.

**3. Luxury's outlier cloud above $900 is a story on its own** : That dense cluster of grey dots stretching to $1,200/mile represents vehicles where the price has completely decoupled from range utility — at that point you're not buying miles, you're buying exclusivity.

**4. Mid-range is where value and quality actually meet** : A median around $210/mile with a compact IQR means Mid-range buyers get a reliable, efficient deal — not as cheap as Budget, but the consistency and build quality step-up makes that gap defensible.

**5. Premium is the segment that needs to justify itself most** : Sitting between Luxury and Mid-range at ~$310/mile, Premium buyers pay significantly more than Mid-range without getting anywhere near Luxury's brand cachet — making it the segment under the most pressure to prove its worth through features and technology.
