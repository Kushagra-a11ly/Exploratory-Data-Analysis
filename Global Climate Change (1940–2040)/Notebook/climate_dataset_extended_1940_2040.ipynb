```python
# 1. Which regions contribute the highest CO₂ emissions?

plt.figure(figsize=(8,5))

region_co2 = df.groupby('Region')['CO2_Emissions_mt'].mean().sort_values()
palette = sns.color_palette("rocket", len(region_co2))

sns.barplot(x=region_co2.values, y=region_co2.index,
            hue=region_co2.index,
            palette=palette,
            legend=False)

plt.title("Average CO₂ Emissions by Region", fontsize=14)
plt.xlabel("CO₂ Emissions (mt)")
plt.ylabel("Region")
plt.show()
```

# CO₂ Emissions Insights

1. **Oceania & Africa Lead Emissions** — Both regions record the highest average CO₂ emissions (~450 mt), significantly outpacing other regions.
2. **Asia is the Lowest Emitter** — Despite being the most populous region, Asia averages the lowest CO₂ emissions (~330 mt), suggesting higher energy efficiency or reporting differences.
3. **Europe & South America are Nearly Identical** — Both regions sit at ~390 mt, indicating similar average emission profiles despite vastly different economic structures.
4. **North America Falls in the Middle** — At ~405 mt, North America sits just above Europe and South America, reflecting moderate-to-high per-region average output.
5. **~120 mt Gap Between Extremes** — The spread between Asia (~330 mt) and Oceania/Africa (~450 mt) reveals a meaningful regional disparity worth targeting in global climate policy.

```python
# 2. Is temperature anomaly increasing with CO₂ emissions?

plt.figure(figsize=(8,5))

n_scenarios = df['Scenario'].nunique()
palette = sns.color_palette("Spectral", n_scenarios)

sns.scatterplot(data=df, x='CO2_Emissions_mt', y='Temp_Anomaly_C',
                hue='Scenario', palette=palette, s=80)

plt.title("CO₂ Emissions vs Temperature Anomaly", fontsize=14)
plt.xlabel("CO₂ Emissions")
plt.ylabel("Temperature Anomaly (°C)")
plt.show()
```

# CO₂ Emissions vs Temperature Anomaly Insights

1. **Green Scenario Keeps Temperatures Lowest** : The Green scenario (teal) clusters heavily between −0.5°C to +1.5°C anomaly, confirming that aggressive emission reduction directly suppresses temperature rise.

2. **BAU Scenario Projects the Most Alarming Future**:Business-As-Usual (yellow-green) spreads furthest right (up to 800+ CO₂) and reaches anomalies of +2.5°C, painting the starkest climate warning.

3. **Positive Correlation Exists Across All Scenarios**: Higher CO₂ emissions consistently align with higher temperature anomalies across Historical, Current, BAU, and Green scenarios alike.

4. **Current (2020–2026) Mirrors Historical Patterns**: Orange (Current) and red (Historical) dots heavily overlap between 100–500 CO₂ range, suggesting emissions trajectories haven't shifted dramatically yet.

5. **Emission Range Widens Drastically in Future Scenarios**: While Historical and Green data stay mostly below 500 CO₂, BAU extends beyond 800, highlighting how inaction exponentially expands the emission ceiling.

```python
# 3. Which countries invest more in renewable energy?

top = df.groupby('Country')['Renewable_Energy_%'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(9,5))

palette = sns.color_palette("crest", len(top))

sns.barplot(x=top.values, y=top.index,
            hue=top.index,
            palette=palette,
            legend=False)

plt.title("Top 10 Countries by Renewable Energy Usage")
plt.xlabel("Renewable Energy %")
plt.ylabel("Country")
plt.show()
```

# Top 10 Countries by Renewable Energy Usage — Insights

1. **All 10 Countries Are Nearly Equal** : Every country sits at ~18% renewable energy usage, making the ranking differences negligibly small.
2. **Germany Leads the Chart** : Germany tops the list, aligning with its well-known Energiewende (energy transition) policy driving renewable adoption.
3. **Diverse Economies, Same Output** : Developed (USA, UK) and developing (Bangladesh, Philippines) nations share identical renewable percentages, suggesting a global adoption plateau.
4. **UAE's Presence is Surprising** : A fossil-fuel-dominant economy like UAE matching Germany in renewables signals a significant strategic shift toward clean energy diversification.
5. **Bangladesh Ranks Last Despite Similar Numbers** : Though at the bottom, Bangladesh achieving ~18% renewables is remarkable given its status as a lower-income developing nation.

```python
# 4. Which regions face the most extreme climate events?

plt.figure(figsize=(8,5))

events = df.groupby('Region')['Extreme_Events'].sum().sort_values()
n_regions = len(events)
palette = sns.color_palette("inferno", n_regions)

sns.barplot(x=events.values, y=events.index,
            hue=events.index,
            palette=palette,
            legend=False)

plt.title("Extreme Climate Events by Region")
plt.xlabel("Number of Events")
plt.ylabel("Region")
plt.show()
```

# Extreme Climate Events by Region Insights

1. **Asia Dominates Extreme Events** : Asia records ~58,000 extreme climate events, nearly double Europe's count, making it by far the most climate-vulnerable region on the planet.
2. **Europe is Second Most Affected** : With ~29,000 events, Europe experiences a surprisingly high frequency, likely driven by escalating heatwaves, floods, and wildfires in recent years.
3. **Africa & North America Are Almost Identical** : Both regions sit at ~14,000 events, suggesting comparable exposure levels despite very different geographic and economic profiles.
4. **Oceania Has the Fewest Events** : At just ~5,000 events, Oceania records the lowest count, though its smaller landmass and population density largely explain the lower absolute numbers.
5. **Asia-Oceania Gap is Staggering** : The difference between Asia (~58,000) and Oceania (~5,000) is over 11x, highlighting extreme regional disparity in climate event frequency and risk exposure.

```python
# 5. How strongly are climate variables correlated?

plt.figure(figsize=(10,6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='icefire', fmt='.2f')
plt.title("Climate Variables Correlation Matrix")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

# Climate Variables Correlation Matrix Insights

1. **Sea Level & Methane Are Almost Perfectly Correlated (0.99)** : These two variables move in near-perfect lockstep over time, strongly suggesting that rising methane emissions are a direct contributor to accelerating sea level rise.
2. **Temperature Anomaly Drives Almost Everything** : Temp_Anomaly_C shows very high correlations with Sea Level (0.98), Methane (0.98), and Extreme Events (0.96), confirming it as the central connecting variable across all major climate indicators.
3. **CO₂ Emissions Are Weakly Correlated With Most Variables** : CO₂_Emissions_mt scores surprisingly low against GDP (0.02), Population (0.07), and Climate Policy Index (0.19), suggesting emissions are not simply explained by economic size or population alone.
4. **GDP & Population Have Minimal Climate Impact Correlations** : Both GDP_Billion and Population_Million show consistently low correlations (0.02–0.32) across all climate variables, debunking the assumption that wealthier or larger nations automatically drive worse climate outcomes.
5. **Renewable Energy & Climate Policy Are Strongly Linked (0.85)** : The high correlation between Renewable_Energy_% and Climate_Policy_Index confirms that stronger climate policies directly translate into greater renewable energy adoption across nations.

```python
# 6. Are stronger climate policies reducing emissions?

plt.figure(figsize=(8,5))
sns.regplot(data=df, x='Climate_Policy_Index', y='CO2_Emissions_mt',
            scatter_kws={'color':'midnightblue'},
            line_kws={'color':'orange'})
plt.title("Policy Strength vs CO₂ Emissions")
plt.xlabel("Climate Policy Index")
plt.ylabel("CO₂ Emissions")
plt.show()
```

# Policy Strength vs CO₂ Emissions Insights

1. **Trend Line Shows a Counterintuitive Positive Slope** : The orange regression line rises from ~360 to ~500 CO₂ as policy strength increases, suggesting stronger policies alone do not immediately reduce emissions — likely due to policy lag effects.
2. **High Emission Outliers Exist Across All Policy Levels** : Dots reaching 700–800 CO₂ are scattered from policy index 10 all the way to 80, indicating that even nations with strong climate policies still produce dangerously high emissions.
3. **Low Emissions Are Concentrated at Extreme Policy Ends** : The lowest CO₂ readings (below 100) appear mostly at both very low and very high policy index values, showing that emission reduction success is inconsistent and non-linear across policy strength.
4. **Data Spread Narrows Slightly at Higher Policy Index** : Beyond index 60–70, the vertical scatter of dots becomes marginally tighter around the 300–500 range, hinting that very strong policies may begin stabilizing emissions within a narrower band.
5. **Weak Policies Show the Most Volatile Emissions** : Between index 10–30, CO₂ values range wildly from near 0 to 700+, revealing that low-policy environments produce the most unpredictable and extreme emission behaviors.

```python
# 7. How has sea level changed over time?

plt.figure(figsize=(9,5))
palette = sns.color_palette("winter")
sea = df.groupby('Year')['Sea_Level_mm'].mean()
plt.plot(sea.index, sea.values, color=palette[3], linewidth=3)
plt.title("Average Sea Level Rise Over Time")
plt.xlabel("Year")
plt.ylabel("Sea Level (mm)")
plt.grid(True)
plt.show()
```

# Average Sea Level Rise Over Time Insights

1. **Rise is Perfectly Linear Since 1940** : The line shows a remarkably straight, uninterrupted climb from 0mm to ~248mm over 100 years, indicating a consistent and relentless rate of sea level rise with zero signs of slowing.
2. **Total Rise of ~248mm in a Century** : From 1940 to 2040, sea levels are projected to rise nearly 250mm (25cm), a significant enough increase to threaten low-lying coastal cities, islands, and millions of people worldwide.
3. **No Policy Intervention is Reflected in the Trend** : The absence of any dip, plateau, or slowdown in the curve suggests that decades of climate agreements and policies have had no measurable impact on halting sea level rise so far.
4. **Rate Averages ~2.5mm Per Year Consistently** : Dividing the total rise (~248mm) across 100 years yields a steady ~2.5mm annual increase, confirming the acceleration concern as even small yearly gains compound into massive long-term consequences.
5. **Post-2020 Projection Continues Unabated** : The 2020–2040 segment shows no deviation from the historical trend, making it a stark forecast that without drastic intervention, the next 20 years will add another ~50mm to already critical sea levels.

```python
# 8. Which scenario produces the highest methane concentration?

plt.figure(figsize=(8,5))

n_scenarios = df['Scenario'].nunique()
palette = sns.color_palette("Pastel1", n_scenarios)

sns.boxplot(data=df, x='Scenario', y='Methane_ppb',
            hue='Scenario', palette=palette, legend=False)

plt.title("Methane Concentration by Scenario")
plt.xlabel("Scenario")
plt.ylabel("Methane (ppb)")
plt.xticks(rotation=15)
plt.show()
```

# Methane Concentration by Scenario Insights

1. **BAU Scenario Has the Highest Median Methane (~1400 ppb)** : The Business-As-Usual box sits highest overall with the widest upper range reaching ~1700 ppb, confirming that inaction leads to the most severe methane concentration outcomes.
2. **Green Scenario Successfully Lowers Methane Levels** : The Green scenario shows the lowest median (~1275 ppb) and a tighter, lower-positioned box, proving that climate-focused interventions meaningfully reduce atmospheric methane concentrations.
3. **Current (2020–2026) Closely Mirrors Historical Patterns** : Both boxes share nearly identical medians (~1350 ppb) and similar spreads, reinforcing that recent years have not yet produced any meaningful departure from long-standing historical methane trends.
4. **BAU Shows the Widest Spread and Highest Uncertainty** : The BAU interquartile range spans ~1250–1550 ppb with whiskers reaching 1700+, indicating that a no-action future carries not just higher methane but also the most unpredictable and volatile outcomes.
5. **Green Scenario Has the Smallest Spread** : The tightest box among all four scenarios belongs to Green, suggesting that strong climate policies not only reduce methane but also stabilize it, creating a more predictable and manageable atmospheric environment.

```python
# 9. Which regions have the highest GDP and emissions together?

plt.figure(figsize=(8,5))

n_regions = df['Region'].nunique()
palette = sns.color_palette("Dark2", n_regions)

sns.scatterplot(data=df, x='GDP_Billion', y='CO2_Emissions_mt',
                hue='Region', palette=palette, s=90)

plt.title("GDP vs CO₂ Emissions by Region")
plt.xlabel("GDP (Billion)")
plt.ylabel("CO₂ Emissions")
plt.show()
```

# GDP vs CO₂ Emissions by Region Insights

1. **No Clear Linear Relationship Between GDP and Emissions** : Despite GDP ranging from $250B to $2800B, CO₂ emissions remain broadly scattered between 100–800 across all regions, proving that higher wealth alone does not predictably drive higher emissions.
2. **Asia Dominates Across the Entire GDP Spectrum** : Orange (Asia) dots are densely spread from low to high GDP ranges with consistently high emissions, confirming Asia as the most emission-intensive region regardless of economic output level.
3. **South America Clusters in the Mid-GDP, Mid-Emission Zone** : Pink (South America) dots concentrate heavily between $500B–$2000B GDP and 200–600 CO₂, suggesting a relatively uniform economic and emission profile across the continent.
4. **North America Reaches the Highest Emission Peaks** : Teal (North America) dots appear prominently at the top of the chart (700–800+ CO₂), indicating that despite fewer data points, North American economies produce some of the most extreme individual emission readings.
5. **Africa Shows Low GDP but Surprisingly Spread Emissions** : Green (Africa) dots sit mostly below $1000B GDP yet their CO₂ values scatter widely from 100 to 700+, suggesting emission levels in Africa are driven by factors beyond just economic size, such as energy mix and industrialization type.

```python
# 10. Which years recorded the highest average temperature anomaly?

temp = df.groupby('Year')['Temp_Anomaly_C'].mean()
plt.figure(figsize=(9,5))
palette = sns.color_palette("cool")
sns.lineplot(x=temp.index, y=temp.values,
             color=palette[4], linewidth=3)
plt.title("Yearly Average Temperature Anomaly")
plt.xlabel("Year")
plt.ylabel("Temperature Anomaly (°C)")
plt.grid(True)
plt.show()
```

# Yearly Average Temperature Anomaly Insights

1. **Temperature Has Risen ~2°C in Just 100 Years** : From −0.27°C in 1940 to +1.77°C by 2040, the total anomaly swing of ~2°C over a single century represents an unprecedented rate of warming in Earth's recorded climate history.
2. **The 1.5°C Paris Agreement Threshold is Already Breached** : The line crosses the critical +1.5°C mark around 2030–2035 and continues rising, signaling that the internationally agreed "safe" warming limit will be surpassed within this decade.
3. **Rate of Warming Visibly Accelerates Post-1980** : The slope of the curve steepens noticeably after 1980, coinciding with rapid industrialization and fossil fuel expansion globally, confirming that human activity has intensified the warming trajectory significantly.
4. **Pre-1960 Period Shows Slight Fluctuation Before Steady Rise** : Between 1940–1960, the line shows minor dips and plateaus around 0°C, representing the last era of relative temperature stability before the uninterrupted upward trend locked in permanently.
5. **Post-2020 Projection Shows No Signs of Plateauing** : The 2020–2040 segment continues its steep climb from ~1.35°C to ~1.77°C with no flattening, making it a stark warning that without dramatic intervention, the world is on track to exceed +2°C well before 2050.

```python
# 11. Which regions are likely to experience rising sea levels?

plt.figure(figsize=(10,6))

n_regions = df['Region'].nunique()
palette = sns.color_palette("tab10", n_regions)

sns.lineplot(data=df, x='Year', y='Sea_Level_mm',
             hue='Region', palette=palette, linewidth=3)

plt.title("Predicted Sea Level Rise by Region")
plt.xlabel("Year")
plt.ylabel("Sea Level (mm)")
plt.grid(alpha=0.3)
plt.show()
```

# Predicted Sea Level Rise by Region Insights

1. **All Regions Show Near-Identical Sea Level Rise Trajectories** : Every region's line overlaps almost perfectly from 1940 to 2040, rising uniformly from 0mm to ~245mm, confirming that sea level rise is a truly global phenomenon affecting all regions equally regardless of geography.
2. **Uncertainty Bands Widen Drastically Post-2000** : The confidence interval shading remains tight before 2000 but expands significantly beyond it, indicating that future sea level predictions carry increasing uncertainty the further out projections extend.
3. **Oceania Carries the Widest Uncertainty Band** : The brown (Oceania) shading spreads the furthest, reaching nearly 280mm on the upper end by 2040, suggesting Oceania faces the highest variability in sea level outcomes and therefore the greatest unpredictability in climate risk planning.
4. **Growth Rate Visibly Accelerates After 1980** : The slope of all regional lines steepens noticeably post-1980, mirroring the temperature anomaly acceleration seen in the same period and reinforcing the direct link between warming temperatures and faster sea level rise.
5. **~245mm Rise Projected Uniformly by 2040** : All six regions converge at approximately 245mm by 2040, a 25cm rise that poses severe threats to coastal infrastructure, island nations, and low-lying delta cities across every part of the world.

```python
# 12.Which variables most strongly predict temperature anomaly?

corr = df.corr(numeric_only=True)
sns.clustermap(corr,
               cmap='vlag',
               annot=True,
               figsize=(10,8),
               linewidths=0.5)
plt.suptitle("Predictors of Temperature Anomaly", y=1.02)
plt.show()
```

# Predictors of Temperature Anomaly Insights

1. **Temp Anomaly, Sea Level & Methane Form a Near-Perfect Cluster (0.98–0.99)** : These three variables are so tightly correlated they essentially move as one unit, confirming that rising temperatures, melting ice driving sea levels, and methane emissions are deeply interconnected feedback loops.
2. **Renewable Energy & Climate Policy Form Their Own Strong Cluster (0.85)** : The dendrogram groups these two together at the top, validating that policy strength and renewable adoption are inseparable twin drivers that must be advanced simultaneously for meaningful climate impact.
3. **CO₂ Emissions Stand Isolated From All Climate Variables** : CO₂_Emissions_mt clusters completely separately with near-zero correlations to GDP (0.02) and Population (0.066), revealing a counterintuitive disconnect that suggests emission levels are driven by structural and behavioral factors rather than pure economic or demographic size.
4. **GDP and Population Are Climate-Irrelevant Variables** : Both GDP_Billion and Population_Million sit in their own isolated cluster with uniformly low correlations (0.02–0.46) across all climate indicators, proving that national wealth and population size are poor predictors of a country's climate footprint or vulnerability.
5. **Year is a Strong Proxy for All Climate Deterioration (0.91–0.94)** : Year correlates highly with Temperature Anomaly (0.93), Sea Level (0.94), and Methane (0.93), meaning time itself is effectively a master predictor of worsening climate conditions — a damning signal that the situation consistently deteriorates with every passing year.

```python
# How do CO₂ emissions vary by Region and Scenario?

result = df.groupby(['Region', 'Scenario']).agg(
    CO2_Avg=('CO2_Emissions_mt', 'mean'),
    Temp_Avg=('Temp_Anomaly_C', 'mean'),
    Events_Total=('Extreme_Events', 'sum')
).reset_index()

print(result)
```

```python
# How does CO₂ emission change across Region vs Scenario?

pivot = df.pivot_table(
    values='CO2_Emissions_mt',
    index='Region',
    columns='Scenario',
    aggfunc='mean'
)

print(pivot)
```
