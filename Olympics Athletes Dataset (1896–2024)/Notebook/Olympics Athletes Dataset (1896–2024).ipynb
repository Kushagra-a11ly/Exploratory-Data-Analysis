```python
# Data Visualization

# 1.Who participates more?

plt.figure(figsize=(6,4))
sns.countplot(x='gender', data=df, hue='gender',
              palette='Set2', legend=False)
plt.title("Gender Distribution of Athletes")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

### Key Insights - Gender Distribution of Athletes 

- The dataset is remarkably balanced — Female (~4,250) and Male (~4,200) athletes are nearly equal in count, with less than a 1% difference between the two groups.
- This near-perfect parity suggests either a deliberately balanced dataset sampling strategy, or a sport/event category where gender representation is structurally equal (e.g. Olympic events with mirrored male/female categories).
- The absence of any third gender category indicates the dataset follows a binary classification — worth noting as a limitation if the analysis aims to be fully inclusive.
- From a modelling perspective, this balanced distribution is ideal — no class imbalance corrections (oversampling, SMOTE, class weights) will be needed for gender as a feature or target variable.

```python
# 2.Which medal is most common?

plt.figure()
df[['gold_medals','silver_medals','bronze_medals']].sum().plot(
    kind='bar', color=['#FFD700','#C0C0C0','#CD7F32']
)
plt.title("Total Medal Distribution")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

### Key Insights: Total Medal Distribution

- 1.Gold medals (~17,200) outnumber Silver (~8,700) and Bronze (~9,100) by nearly 2:1 — a striking imbalance that suggests the dataset counts individual athlete medals rather than unique event outcomes, where gold, silver, and bronze should theoretically be equal in count.
- 2.Silver and Bronze are nearly identical in distribution (~8,700 vs ~9,100), differing by less than 5% — the real anomaly is gold's disproportionate dominance.
- 3.The gold inflation likely reflects team sports where every squad member receives a gold medal per event win, while silver and bronze follow the same logic but with fewer team-based events in the dataset.
- 4.Alternatively, this could indicate data quality issue — duplicate gold medal entries or multi-counting across relay/team disciplines — worth investigating before using medal count as a feature in any model.
- 5.For analysis purposes, treating total medals (gold + silver + bronze) as a single success metric may be more reliable than weighting by medal type, given this imbalance.

```python
# 3.What is the peak athlete age?

plt.figure()
sns.histplot(df['age'], kde=True, color='skyblue', bins=30)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()
```

### Key Insights: Age Distribution of Athletes


- 1.The distribution is remarkably uniform between ages 15–41, with most bins holding steady at 280–320 athletes — an unusually flat pattern rarely seen in real-world athletic data, which typically peaks in the mid-20s and tapers off.
- 2.Two visible dips occur around ages 23–24 and 28–30, breaking the otherwise flat plateau — these could reflect natural career transition points (post-junior → senior circuit, or mid-career breaks) or simply gaps in data collection.
- 3.The KDE (smoothed line) confirms the near-flat trend with a slight dip in the mid-20s and a sharp drop-off after 41, where athlete representation falls steeply — consistent with physical retirement age across most sports.
- 4.The presence of athletes as young as 15 indicates inclusion of youth-eligible sports such as gymnastics, swimming, or figure skating, where peak competitive age is significantly lower than contact or strength-based sports.
- 5.The flatness of this distribution is a strong signal of deliberate dataset balancing across age groups — natural Olympic data would show a clear right-skewed peak around 22–26. This should be factored in when drawing population-level conclusions.

```python
# 14.Are medals increasing over years?

plt.figure()
df.groupby('year')['total_medals_won'].sum().plot(color='red', marker='o')
plt.title("Medals Over Time")
plt.xlabel("Year")
plt.ylabel("Total Medals")
plt.show()
```

### Key Insights: Medals Over Time (1896–2024)

- 1.A dramatic structural break occurs between 1920 and 1932 — total medals plummet from ~1,500 to ~500, almost certainly reflecting the cancellation of the 1916 Games (WWI) and 1940/1944 Games (WWII), combined with a reduced event programme in the early post-war era rather than a genuine decline in athletic output.
- 2.The 1896–1920 period shows a consistently high medal count (~1,250–1,500), which is counterintuitive given the Olympics were in their infancy — suggesting either dataset overcounting of early Games, inflation from multi-event athletes, or inclusion of demonstration sports that were later excluded.
- 3.A long flat plateau persists from 1932–1988 (~500–650 medals), representing the stable mid-century Olympic programme before the major expansion of events and participating nations in the modern era.
- 4.Post-1992 data shows extreme volatility — alternating spikes between ~450 and ~1,500 in rapid succession, a pattern inconsistent with real Olympic medal growth which is gradual. This oscillation strongly suggests Summer and Winter Games are being plotted on the same axis without differentiation, causing the alternating high/low pattern as the two Games have very different total medal counts.
- 5.The most critical analytical action for time-series work on this dataset is to split Summer and Winter Olympics into separate series — combining them on a single timeline produces a misleading sawtooth pattern that obscures genuine longitudinal trends in either Games' medal evolution.
