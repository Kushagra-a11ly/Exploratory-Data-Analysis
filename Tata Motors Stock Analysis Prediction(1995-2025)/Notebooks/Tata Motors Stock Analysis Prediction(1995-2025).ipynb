```python
# Which variables are strongly related to each other?

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# correlation
corr = df.corr(numeric_only=True)

# remove redundant upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(10, 6))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)

plt.title("Correlation Heatmap", fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Insights:

# 1.Price variables are highly redundant — they all move almost identically.
# 2.Drawdown and Peak are just derived from price, not independent signals.
# 3.Liquidity metrics (Volume, Turnover, Trades) form a separate cluster.
# 4.Daily returns are mostly independent and highly noisy.
# 5.Month shows no meaningful correlation — weak seasonality.
# 6.Year shows moderate correlation — indicating an overall trend over time.
```

```python
# How do stock returns vary across months and years?

import seaborn as sns
import matplotlib.pyplot as plt

df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year

pivot = df.pivot_table(
    values='Daily_Return_%',
    index='Year',
    columns='Month',
    aggfunc='mean'
)

pivot = pivot * 100
plt.figure(figsize=(12,6))
sns.heatmap(
    pivot,
    cmap='coolwarm',
    annot=True,
    fmt=".2f"
)
plt.title("Monthly Average Returns (%)")
plt.xlabel("Month")
plt.ylabel("Year")
plt.show()
```

```python
# Insights:

# 1.September 2011 stands out with an extreme loss of -333.97%, the single worst monthly return in the entire dataset, likely reflecting the global market crash and eurozone crisis impact on TATAMOTORS.
# 2.October 2019 recorded the highest gain at +226.40%, followed closely by October 2021 (+202.52%) and April 2020 (+260.21%), suggesting October and April are historically strong months for the stock.
# 3.Month 2 (February) and Month 3 (March) show frequent blue (negative) tones across multiple years, making early calendar months structurally weak periods with recurring losses, particularly February 2020 (-164.18) and March 2020 (-181.33).
# 4.January shows a strong bullish bias with several high positive returns — 2012 (+144.10), 2021 (+188.23), and 2015 (+82.11) — making it one of the most consistently red (positive) months across the heatmap.
# 5.2020 was the most volatile year overall, displaying both the deepest losses (Feb: -164.18, Mar: -181.33) and strong recoveries (Apr: +260.21, Aug: +144.24, Nov: +158.58), directly reflecting COVID-19 crash and subsequent stimulus-driven rebound.
```

```python
# How much loss (from peak) did the investment experience over time?

df['Cumulative'] = (1 + df['Daily_Return_%']/100).cumprod()
df['Peak'] = df['Cumulative'].cummax()
df['Drawdown'] = df['Cumulative'] - df['Peak']

plt.figure(figsize=(10,5))
plt.plot(df['Date'], df['Drawdown'], color='red')
plt.title("Drawdown Over Time")
plt.show()
```

```python
# Insights:

# 1.Maximum drawdown hits nearly -85% in the earliest period, representing a catastrophic peak-to-trough decline that wiped out most investor wealth — the worst capital destruction phase in TATAMOTORS' observed history.
# 2.A prolonged recovery phase spans the mid-section, where drawdown gradually improves from -80% to -40%, suggesting a slow, multi-year climb back toward previous highs with multiple failed recovery attempts visible as dips within the trend.
# 3.Two distinct bottom formations (~-80% and ~-85%) appear early and mid-chart, indicating the stock retested its lows twice before mounting a sustained recovery — a classic double-bottom structure signaling eventual trend reversal.
# 4.The most recent period shows a near-zero drawdown recovery, meaning the stock approached or hit new all-time highs, followed by a sharp fresh drawdown to ~-50%, suggesting a significant peak was formed recently and the stock is currently in a new corrective phase.
# 5.Drawdown never sustained zero (new highs) for extended periods, indicating TATAMOTORS spends most of its time below previous peaks — characteristic of a high-volatility, cyclical stock that rewards only precise entry-exit timing rather than passive long-term holding.
```
