```python
# Data Cleaning and Preprocessing

# Check for missing values
df.isnull().sum()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df['Cancelled Rides by Customer'] = df['Cancelled Rides by Customer'].fillna(0)
df['Cancelled Rides by Driver'] = df['Cancelled Rides by Driver'].fillna(0)

df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].fillna('Not Cancelled')
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].fillna('Not Cancelled')

df['Incomplete Rides'] = df['Incomplete Rides'].fillna(0)
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].fillna('Completed')

df['Driver Ratings'].fillna(df['Driver Ratings'].mean(), inplace=True)
df['Customer Rating'].fillna(df['Customer Rating'].mean(), inplace=True)
```

```python
df['Booking Value'] = df['Booking Value'].fillna(0)
df['Ride Distance'] = df['Ride Distance'].fillna(0)
df['Payment Method'] = df['Payment Method'].fillna('Not Applicable')
df['Avg CTAT'] = df['Avg CTAT'].fillna(df['Avg CTAT'].median())
df['Avg VTAT'] = df['Avg VTAT'].fillna(df['Avg VTAT'].median())
```

```python
df.isnull().sum()
```

```python
# What % of rides are completed vs cancelled?

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.countplot(
    x='Booking Status',
    data=df,
    color='skyblue'
)
plt.title('Booking Status Distribution')
plt.xticks(rotation=30)
plt.xlabel('Booking Status')
plt.ylabel('Count')
plt.show()
```

# Insights: Booking Status Distribution

**1. Completed rides are clearly winning**
With ~93K completions, the platform is doing its core job well — but the real story is in everything else that's going wrong around it.

**2. Drivers are cancelling way too often**
~27K driver cancellations is a serious red flag. When drivers bail that frequently, it's usually a sign they're cherry-picking rides or the incentives just aren't working.

**3. ~11K rides found absolutely no driver**
That's not a rounding error — that's a supply problem. Certain areas or peak hours are clearly being underserved and customers are just left hanging.

**4. Incomplete rides are low, but don't ignore them**
~9–10K incomplete bookings sounds small against 93K completions, but each one is a broken experience worth digging into before it quietly grows.

**5. Customer cancellations and "no driver found" are probably the same problem**
When both sit around ~11K, it's hard not to wonder if customers aren't actually cancelling out of frustration from waiting too long — which means fixing driver supply could knock out two problems at once.

If drivers are having to cancel because customers show up in oversized groups,

~6–7K cancellations because a customer was coughing or sick is no longer just

**1. Distance alone doesn't determine price and this chart proves it:**
Rides of the same distance (say, 10km) can cost anywhere from ₹0 to ₹4000+,

heavy lifting in fare calculation.

**2. Short rides have the widest price variance  that's counterintuitive:**
You'd expect short trips to be cheap and predictable, but the scatter near 
0–10km is just as spread out as longer trips, suggesting premium vehicle 
bookings on short distances are quite common.

**3. There's a visible revenue floor around ₹500–₹1000 regardless of distance:**
Across almost every distance band, there's a dense cluster sitting between 
₹500–₹2000 — this is likely the base fare band that the pricing model 
consistently lands in for standard rides.

**4. High-value rides (₹3000+) happen at every distance, not just long ones:**
Seeing ₹3000–₹4000 bookings at under 5km distance is a clear signal that 
premium vehicle categories like Uber XL or Premier Sedan are being booked 
for very short trips — convenience and comfort over cost.

**5. The scatter gets thinner at 40–50km but doesn't disappear:**
Long-distance rides are rare (as expected), but they still cover the full 
revenue range — meaning even long trips aren't guaranteed to be high revenue, 
which points to a possible pricing inefficiency at higher distances.

If 15 out of every 100 drivers are in the 2–4 range and the platform is doing 
tens of thousands of rides, that translates to a significant volume of 
disappointing customer experiences happening quietly every single day.

# Insights: Correlation Heatmap

**1. Booking Value and Ride Distance have the only meaningful correlation at 0.42:**
A moderate positive link makes intuitive sense — longer rides cost more — but 
0.42 also means distance explains less than half the fare variation, confirming 
that other factors like vehicle type and surge are doing significant pricing work.

**2. Ratings have virtually zero correlation with anything — and that's alarming:**
Driver and Customer ratings show correlations of essentially 0.000 with both 
fare and distance, meaning how much you pay or how far you travel has absolutely 
no bearing on the rating either party gives — satisfaction is driven by 
something this dataset isn't capturing.

**3. Driver Ratings and Customer Ratings don't correlate with each other either:**
At -0.001, drivers and customers are rating completely independently — a 
5-star driver doesn't reliably get a 5-star customer, and vice versa, suggesting 
both parties are evaluating entirely different aspects of the same ride.

**4. The near-zero correlations aren't noise — they're a signal to look elsewhere:**
When four variables barely interact, it tells you the real drivers of 
performance (cancellations, wait times, location, time of day) are missing 
from this correlation matrix entirely and need to be added for deeper analysis.

**5. This heatmap is essentially a two-variable story dressed up as four:**
Strip away the self-correlations and the only actionable number here is 0.42 — 
everything else is statistical silence, which means the dataset needs richer 
features before correlation analysis can guide any real business decisions.
