```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
import plotly.express as px, streamlit as st
sns.set(style="whitegrid", font_scale=1.1)

df = pd.read_csv("ultimate_games_dataset.csv")
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
```

```python
# 1. Genre Dominance Analysis

import matplotlib.pyplot as plt

plt.rcdefaults()

plt.figure(figsize=(12,6))
g = df['all_genres'].value_counts().head(10)

sns.barplot(
    x=g.values,
    y=g.index,
    hue=g.index,
    palette='viridis',
    legend=False,
    edgecolor='black'
)

plt.title("Top 10 Game Genres")
plt.xlabel("Game Count")
plt.ylabel("Genres")
plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.show()
```

Insights - Top 10 Game Genres:
- **Adventure|Indie** is the most dominant genre combination, showing strong demand for creative exploration-based games.
- **Casual|Indie** closely follows, highlighting the popularity of accessible and relaxing gameplay experiences.
- Hybrid genres outperform standalone categories, indicating players prefer more diverse gaming experiences.
- **Adventure** remains one of the strongest core genres due to its broad audience appeal.
- Action-based genres continue to perform consistently across multiple genre combinations.
- **Adventure|Casual|Indie** demonstrates growing interest in narrative-driven casual gaming.
- Standalone Casual games rank lowest among the top 10, suggesting better performance when paired with other genres.
- **Action|Shooter** maintains solid popularity but trails broader action-oriented genre mixes.

```python
# 2. Game Releases Trend

plt.figure(figsize=(13,5))
y = df['release_year'].value_counts().sort_index()
sns.lineplot(x=y.index, y=y.values, color='crimson', linewidth=3)
plt.title("Game Releases Over Time", fontsize=16, weight='bold')
plt.xlabel("Release Year"); plt.ylabel("Games Released")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Game Releases Over Time — Key Insights

**1. Dormant Era (1978–1999):** Release volumes remained negligible across two decades, indicative of high entry barriers, limited distribution channels, and a predominantly hardware-driven market controlled by a few major publishers.

**2. Growth Phase (2000–2012):** A consistent upward trajectory emerged as digital storefronts dismantled traditional retail gatekeeping, enabling smaller studios to reach global audiences without physical distribution infrastructure.

**3. Market Saturation Peak (2015–2016):** The spike to ~1,600 annual releases represents an inflection point driven by Steam Greenlight's low-barrier submission model, which temporarily overwhelmed curation mechanisms and flooded the market with indie titles.

**4. Correction & Secondary Surge (2017–2022):** Post-peak consolidation reflects platform-level quality controls tightening, while the secondary rise to ~1,050 releases around 2021–2022 correlates with pandemic-driven consumer demand and a surge in remote development activity.

**5. Trailing Data Artifact (2023–Present):** The sharp drop toward zero is not indicative of market contraction; rather, it reflects dataset recency lag — a common anomaly in release databases where recent entries remain incompletely catalogued at the time of data extraction.

```python
# 3. User Rating Distribution
plt.figure(figsize=(10,5))
sns.histplot(df['user_rating'], bins=30, kde=True,
             color='darkorange', edgecolor='black')
plt.title("User Rating Distribution", fontsize=16, weight='bold')
plt.xlabel("User Rating"); plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```

## User Rating Distribution — Key Insights

**1. Left-Skewed Distribution:** The distribution is negatively skewed, with the majority of ratings concentrated between 3.5–4.0, indicating users tend to rate games favorably rather than critically.

**2. Modal Peak at 3.75:** The highest frequency bar (~1,750) falls around the 3.75 rating mark, establishing this as the most commonly assigned score across the entire dataset.

**3. Thin Lower Tail (1.0–2.0):** Ratings below 2.0 are sparse and infrequent, suggesting either a genuine absence of poor-quality titles or a positivity bias where dissatisfied users abstain from rating altogether.

**4. Healthy Mid-Range Engagement (2.5–3.5):** A gradual and consistent frequency build-up across this range reflects a broad base of average-to-good titles, supporting a well-distributed review ecosystem without extreme polarization.

**5. Sharp Right Cutoff Post-4.0:** Frequency drops steeply beyond 4.0 despite the scale extending to 5.0, implying that near-perfect scores are rare and reserved, lending credibility and weight to high-rated titles in this dataset.

```python
# 4. Critic vs User Ratings

plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x='metacritic', y='user_rating',
                hue='is_multiplayer', palette='Set2', edgecolor='black')
plt.title("Metacritic vs User Rating", fontsize=16, weight='bold')
plt.xlabel("Metacritic Score"); plt.ylabel("User Rating")
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

## Metacritic vs User Rating — Key Insights

**1. Positive Correlation (Weak-to-Moderate):** A visible upward trend confirms that higher Metacritic scores generally align with better user ratings, though the wide vertical spread indicates critic and user sentiment frequently diverge at the individual game level.

**2. High Variance in Mid-Range Scores (50–75):** The densest and most scattered region lies between Metacritic scores of 50–75, reflecting the highest disagreement between critics and users — a zone where subjective player experience diverges most from professional critique.

**3. Single-Player Titles Dominate the Dataset:** Green points (is_multiplayer = 0) significantly outnumber orange (is_multiplayer = 1), suggesting the dataset is heavily weighted toward single-player experiences, which may bias overall rating distributions.

**4. Multiplayer Games Cluster in Mid-to-High Critic Scores:** Orange points (multiplayer titles) are predominantly concentrated between Metacritic scores of 60–90, implying multiplayer games are less likely to receive extremely low or extremely high critical assessments.

**5. User Ratings Rarely Reflect Critic Extremes:** Even for games scoring 90+ on Metacritic, user ratings cap around 4.5–5.0 with considerable spread, reinforcing that critical acclaim does not guarantee universal player satisfaction.

```python
# 5. Multiplayer Impact

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='is_multiplayer',
    y='user_rating',
    hue='is_multiplayer',
    palette='coolwarm',
    linewidth=2,
    legend=False
)

plt.title("Multiplayer Effect on Ratings")
plt.xlabel("Multiplayer")
plt.ylabel("User Rating")
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

## Multiplayer Effect on Ratings — Key Insights

**1. Single-Player Games Rate Higher Overall:** Single-player titles (0) exhibit a median rating of ~3.5 compared to ~3.2 for multiplayer (1), suggesting users consistently reward solo experiences with marginally higher satisfaction scores.

**2. Multiplayer Games Show Greater Rating Volatility:** The interquartile range (IQR) for multiplayer titles is visibly wider (~2.7–3.75), indicating higher inconsistency in user satisfaction — likely driven by server quality, matchmaking experience, and community behavior factors beyond core gameplay.

**3. Single-Player Lower Fence Indicates Outlier Concentration:** The presence of multiple outlier points below 1.5 for single-player games suggests a small but notable segment of critically poor-performing titles that skew perception without materially affecting the median.

**4. Upper Whiskers Are Comparable Across Both Groups:** Both categories top out near 4.75–5.0, confirming that exceptional user experiences are achievable regardless of multiplayer mode — excellence is format-agnostic.

**5. Multiplayer Median Falls Below Single-Player Q1:** The median of multiplayer games (~3.2) sits at or below the first quartile of single-player games (~3.0), implying that the majority of multiplayer titles perform in a range where single-player games are still considered below average.

```python
# IQR Method

Q1 = df['popularity_score'].quantile(0.25)
Q3 = df['popularity_score'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['popularity_score'] < lower) |
    (df['popularity_score'] > upper)
]

outliers
```
